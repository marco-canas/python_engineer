Este es un programa de formación estructurado en **4 niveles**. Está diseñado para llevarte desde la sintaxis básica hasta la arquitectura de sistemas, que es lo que realmente separa a un programador de un **Python Engineer**.

---



## Programa: Python Engineering Professional Path



### Nivel 1: El Núcleo del Lenguaje (Fundamentos Sólidos)
*Objetivo: Dominar la sintaxis y la lógica sin depender de tutoriales.*

* **Tipado y Estructuras:** Uso avanzado de listas, diccionarios (comprensiones), sets y tuplas.
* **Programación Funcional:** Funciones lambda, `map`, `filter`, y el uso de *closures*.
* **Programación Orientada a Objetos (OOP):** Clases, herencia, mixins y métodos mágicos (`__init__`, `__str__`, `__call__`).
* **Gestión de Entornos:** Configuración profesional con `pyenv`, `poetry` o `venv`.



### Nivel 2: Ingeniería y Calidad de Código
*Objetivo: Escribir código que sea mantenible y escalable en producción.*

* **Patrones de Diseño:** Singleton, Factory y Observer aplicados en Python.
* **Manejo de Errores:** Jerarquía de excepciones y creación de excepciones personalizadas.
* **Testing:** Pruebas unitarias con `pytest`, uso de *fixtures* y *mocking*.
* **Tipado Estático:** Implementación de `mypy` y *type hinting* para evitar errores en tiempo de ejecución.
* **Clean Code:** Estándares PEP 8, linters (`Ruff`, `Flake8`) y formateadores (`Black`).



### Nivel 3: Conectividad y Datos
*Objetivo: Hacer que tu aplicación hable con el mundo exterior.*

* **Bases de Datos:** SQL avanzado (PostgreSQL), uso de ORMs (`SQLAlchemy` o `Django ORM`) y migraciones con `Alembic`.
* **Asincronía (AsyncIO):** Programación no bloqueante, `async/await` y concurrencia.
* **APIs:** Construcción de servicios con **FastAPI** (preferido por rendimiento) o **Django Rest Framework**.
* **Consumo de Recursos:** Manejo de peticiones HTTP, protocolos de autenticación (OAuth2, JWT).



### Nivel 4: Despliegue y Ecosistema (Professional DevOps)
*Objetivo: Llevar tu código de tu computadora a la nube.*

* **Contenedores:** Dockerización de aplicaciones Python (creación de `Dockerfiles` optimizados).
* **CI/CD:** Automatización de pruebas y despliegue con GitHub Actions.
* **Cloud:** Despliegue básico en AWS (Lambda, EC2) o Google Cloud.
* **Monitoreo:** Logging profesional y manejo de métricas básicas.

```python
import os

def crear_estructura_curso():
    # Ruta base solicitada
    base_path = r"C:\Users\marco\OneDrive - Universidad de Antioquia\Documentos\1_administracion\nuevo_empleo\python_engineer"
    
    # Definición de la estructura: { Carpeta: [Archivos] }
    estructura = {
        "01_fundamentos_core": [
            "01_tipado_y_estructuras.ipynb",
            "02_programacion_funcional.py",
            "03_poo_avanzada.py",
            "04_gestion_entornos_poetry.txt"
        ],
        "02_ingenieria_y_calidad": [
            "01_patrones_diseno.py",
            "02_manejo_excepciones.py",
            "03_testing_pytest.py",
            "04_estatic_typing_mypy.py",
            "05_clean_code_pep8.md"
        ],
        "03_conectividad_y_datos": [
            "01_sql_y_postgresql.ipynb",
            "02_orm_sqlalchemy.py",
            "03_asyncio_concurrencia.py",
            "04_fastapi_rest_basics.py"
        ],
        "04_devops_y_despliegue": [
            "01_dockerizacion_apps.ipynb",
            "02_github_actions_cicd.yml",
            "03_cloud_basics_aws.md"
        ],
        "05_proyecto_final_scraper": [
            "app.py",
            "scraper.py",
            "database.py",
            "requirements.txt",
            "Dockerfile"
        ],
        "recursos_y_notas": [
            "glosario_terminos.md",
            "enlaces_utiles.md"
        ]
    }

    print(f"--- Iniciando creación de estructura en: {base_path} ---")

    # Crear la carpeta base si no existe
    if not os.path.exists(base_path):
        try:
            os.makedirs(base_path)
            print(f"Carpeta base creada.")
        except Exception as e:
            print(f"Error al crear la carpeta base: {e}")
            return

    # Crear subcarpetas y archivos
    for carpeta, archivos in estructura.items():
        ruta_carpeta = os.path.join(base_path, carpeta)
        os.makedirs(ruta_carpeta, exist_ok=True)
        print(f"Directorio listo: {carpeta}")
        
        for archivo in archivos:
            ruta_archivo = os.path.join(ruta_carpeta, archivo)
            if not os.path.exists(ruta_archivo):
                with open(ruta_archivo, 'w', encoding='utf-8') as f:
                    # Agregar un pequeño encabezado a cada archivo
                    if archivo.endswith('.py'):
                        f.write(f"# Ejercicios de {archivo.replace('_', ' ')}\n")
                    elif archivo.endswith('.ipynb'):
                        f.write('{"cells": [], "metadata": {}, "nbformat": 4, "nbformat_minor": 5}')
                print(f"  - Archivo creado: {archivo}")
            else:
                print(f"  - El archivo {archivo} ya existe, saltando...")

    print("\n--- ¡Estructura completada con éxito! ---")
    print("Sugerencia: Abre esta carpeta en VS Code para comenzar tu camino.")

if __name__ == "__main__":
    crear_estructura_curso()
```
