## Fase 1: El Core de Python (Días 1-2)
Wompi utiliza Python para procesos de backend complejos. No basta con que funcione; debe ser "Pythonic".

* **Asincronismo:** Domina `asyncio`. En las pasarelas de pagos, las llamadas a APIs externas son constantes; saber manejar la concurrencia es vital.
* **Tipado Dinámico vs. Estático:** Practica el uso de **Type Hinting** (librería `typing`). Un código profesional en Wompi debe estar debidamente tipado.
* **Gestión de Dependencias:** Asegúrate de entender `Poetry` o `Pipenv`, que son estándar en entornos modernos de desarrollo.
* **Pydantic:** Es fundamental para la validación de datos en APIs.

---



## Fase 2: Backend y Arquitectura (Días 3-4)
Siendo una Fintech, la estructura de la aplicación es sagrada.

* **FastAPI / Flask:** Son los frameworks más probables. Practica creando una API que reciba un "pago", lo valide y lo guarde.
* **Patrones de Diseño:** Repasa **Domain-Driven Design (DDD)**. Wompi separa muy bien la lógica de negocio de la infraestructura.
* **Bases de Datos:** Practica consultas complejas en **PostgreSQL** y el uso de ORMs como `SQLAlchemy`. Entiende cómo manejar **transacciones** (Atomicidad) para que un pago no se cobre dos veces o se pierda.

---



## Fase 3: Calidad y DevOps (Día 5)
En una empresa de este calibre, si no tiene pruebas, no existe.

* **Testing:** Dominio total de `pytest`. Practica **Unit Testing**, **Integration Testing** y el uso de **Mocks** (simular una respuesta de un banco sin llamar al banco real).
* **Docker:** Debes ser capaz de "dockerizar" tu aplicación. Crea un `Dockerfile` optimizado y un `docker-compose.yml` para levantar tu app y tu DB.
* **Principios SOLID:** Tu código será revisado por seniors. Asegúrate de que cada clase y función tenga una única responsabilidad.

---



## Fase 4: El "Simulacro Wompi" (Días 6-7)
Dedica el fin de semana a un proyecto pequeño pero matador que replique su entorno:

1.  **El Reto:** Crea un microservicio que reciba una orden de compra.
2.  **Requisitos:**
    * Usa **FastAPI**.
    * Valida la entrada con **Pydantic**.
    * Guarda en **PostgreSQL** usando un contenedor de Docker.
    * Implementa un **Webhook** (simulado) que notifique cuando el "pago" fue exitoso.
    * Logra una cobertura de pruebas (Coverage) superior al **80%**.

---



## Consejos "Pro" para Wompi
* **Seguridad:** Menciona que conoces sobre el estándar **PCI DSS** (aunque no seas experto, saber que existe para proteger datos de tarjetas suma puntos).
* **Cultura:** Wompi valora la proactividad. Durante la semana de prueba, no solo resuelvas el ticket; pregunta por el contexto y propón mejoras de rendimiento.
* **Clean Code:** Usa `black`, `isort` y `flake8` para que tu estilo de código sea impecable desde el primer commit.
