### Celda 1: Introducción y Lambdas

```python
# PROGRAMACIÓN FUNCIONAL: MENOS BUCLES, MÁS DECLARATIVIDAD
# Las funciones Lambda son el bloque básico. Son rápidas y anónimas.

from typing import List, Callable
from functools import reduce

# Sintaxis: lambda argumentos: expresion
saludo = lambda nombre: f"Hola, {nombre}, bienvenido al curso de Ingeniería."

print(saludo("Marco"))

# Ejemplo práctico: Ordenar una lista de tuplas por el segundo elemento (la edad)
usuarios = [("Marco", 30), ("Elena", 25), ("Lucas", 35)]
usuarios_ordenados = sorted(usuarios, key=lambda x: x[1])

print(f"Usuarios ordenados por edad: {usuarios_ordenados}")
```

### Celda 2: El poder de MAP (Transformación)


```python
# MAP: Aplica una transformación a cada elemento. 
# Importante: Map devuelve un iterador, lo que ahorra memoria.

precios_usd: List[float] = [10.99, 45.50, 100.0, 5.25]

# Transformación: Aplicar un impuesto del 19% (IVA)
precios_con_iva = list(map(lambda p: round(p * 1.19, 2), precios_usd))

print(f"Precios originales: {precios_usd}")
print(f"Precios con IVA:    {precios_con_iva}")
```

### Celda 3: FILTER (Limpieza de datos)


```python
# FILTER: Deja pasar solo los elementos que cumplen una condición.

log_niveles = ["INFO", "ERROR", "DEBUG", "ERROR", "WARNING", "INFO", "ERROR"]

# Filtrar solo los mensajes críticos (ERROR)
solo_errores = list(filter(lambda log: log == "ERROR", log_niveles))

print(f"Total de registros: {len(log_niveles)}")
print(f"Registros de error: {solo_errores}")
print(f"Cantidad de errores detectados: {len(solo_errores)}")
```

### Celda 4: REDUCE (Consolidación)


```python
# REDUCE: Toma todos los elementos y los "reduce" a un solo valor (suma, producto, etc.)

ventas_dia: List[int] = [150, 300, 450, 100]

# Sumar todas las ventas usando reduce
total_ventas = reduce(lambda acumulador, actual: acumulador + actual, ventas_dia)

print(f"El total de ventas del día es: ${total_ventas}")
```

### Celda 5: Reto de Ingeniería (Combinando todo)


```python
# RETO: Procesamiento de datos encadenado.
# Imagina que recibes estas edades de una base de datos:
edades = [12, 25, 18, 40, 15, 30, 55, 8]

# TAREA: 
# 1. Filtra solo las edades que sean mayores o iguales a 18 (Adultos).
# 2. A esas edades, súmales 1 año (Simulando un cierre de año).
# 3. Calcula el promedio de esas edades resultantes usando reduce.

# Escribe tu solución aquí:
adultos = list(filter(lambda e: e >= 18, edades))
adultos_proximo_ano = list(map(lambda e: e + 1, adultos))
suma_total = reduce(lambda a, b: a + b, adultos_proximo_ano)
promedio = suma_total / len(adultos_proximo_ano)

print(f"Edades procesadas: {adultos_proximo_ano}")
print(f"Promedio de edad de adultos el próximo año: {promedio}")
```
