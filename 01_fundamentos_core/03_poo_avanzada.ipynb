## 1. Encapsulamiento (Protección de la Integridad)

**Definición:** Es la ocultación de los detalles internos de un objeto para asegurar que su estado solo cambie de formas permitidas. En Python, usamos el prefijo `__` para atributos privados y el decorador `@property` para exponer datos de manera controlada.



**Aplicación en Geometría:** Un `Círculo` no puede tener un radio negativo. El encapsulamiento evita que un usuario asigne un valor ilógico.




````python
import math

class Circulo:
    def __init__(self, radio: float):
        self.__radio = radio  # Atributo privado (con __)

    @property
    def radio(self) -> float:
        """Getter para el radio."""
        return self.__radio

    @radio.setter
    def radio(self, valor: float):
        """Setter con validación."""
        if valor < 0:
            raise ValueError("""En geometría euclidiana, 
                             el radio no puede ser negativo.""")
        self.__radio = valor

    @property
    def area(self) -> float:
        """El área como propiedad (sin parámetros)."""
        return math.pi * (self.__radio ** 2)
````

````python

# Ejemplo:
c = Circulo(3) # Creación de un círculo con radio 5
````

````python
c.radio = 2  # Usa el setter con validación
````

````python
print(f"Área: {c.area:.2f}")
````


## 2. Herencia y Composición (Relaciones Lógicas)



**Definición:** * **Herencia ("Es un"):** Una clase deriva de otra para reutilizar comportamiento (ej. un Cuadrado *es un* Polígono).
* **Composición ("Tiene un"):** Una clase se construye usando otras clases (ej. un Triángulo *tiene tres* Puntos).

**Aplicación en Geometría:** Un polígono se define por sus vértices (Composición de puntos).




Para completar estas clases como un **Python Engineer**, debemos aplicar la **Geometría Analítica**. La clave aquí es la fórmula de la distancia entre dos puntos en el plano cartesiano:

$$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

Vamos a implementar esto de forma elegante, agregando un método de utilidad en la clase `Punto` y completando la lógica del `Triangulo`.

---



## Código Completado y Profesional


````python
import math
from typing import List

class Punto:
    def __init__(self, x: float, y: float):
        self.x = x
        self.y = y

    def distancia_a(self, otro: 'Punto') -> float:
        """Calcula la distancia euclidiana hacia otro punto."""
        return math.sqrt((otro.x - self.x)**2 + (otro.y - self.y)**2)

    def __repr__(self) -> str: 
        # Este método es para que al imprimir un punto, se vea algo legible
        return f"Punto({self.x}, {self.y})"
````

````python

class Figura(Punto):
    def __init__(self, nombre: str):
        self.nombre = nombre
    
    def calcular_area(self):
        """Método abstracto (para ser definido en hijas)."""
        pass
````

````python

class Triangulo(Figura):
    def __init__(self, p1: Punto, p2: Punto, p3: Punto):
        super().__init__("Triángulo")
        # Composición: El triángulo TIENE puntos
        self.vertices: List[Punto] = [p1, p2, p3] 
    
    def perimetro(self) -> float:
        """Suma las distancias entre sus 3 vértices."""
        p1, p2, p3 = self.vertices
        lado_a = p1.distancia_a(p2)
        lado_b = p2.distancia_a(p3)
        lado_c = p3.distancia_a(p1)
        return lado_a + lado_b + lado_c

    def area(self) -> float:
        """Calcula el área usando la Fórmula de Herón."""
        p1, p2, p3 = self.vertices
        a = p1.distancia_a(p2)
        b = p2.distancia_a(p3)
        c = p3.distancia_a(p1)
        s = (a + b + c) / 2  # Semiperímetro
        return math.sqrt(s * (s - a) * (s - b) * (s - c))

    def __str__(self) -> str:
        return f"{self.nombre} con vértices en {self.vertices}"
````

````python
Punto1 = Punto(0, 0)
Punto2 = Punto(4, 0)    
Punto3 = Punto(4, 3)
triangulo = Triangulo(Punto1, Punto2, Punto3)
print(triangulo)
print(f"Perímetro: {triangulo.perimetro():.2f}")
print(f"Área: {triangulo.area():.2f}")
````

## 3. Métodos de Clase y Estáticos

**Definición:**
* **`@classmethod`:** Recibe la clase (`cls`) como argumento. Útil para "fábricas" que crean objetos de formas alternativas.
* **`@staticmethod`:** Funciones que viven dentro de la clase por organización, pero no acceden a datos del objeto o la clase.

**Aplicación en Geometría:** Crear un cuadrado a partir de un solo lado (fábrica) o validar si tres puntos pueden formar un triángulo (estático).


````python
class Cuadrado:
    def __init__(self, lado: float):
        self.lado = lado

    @classmethod
    def desde_area(cls, area: float):
        """Crea un cuadrado calculando el lado a partir del área."""
        lado = math.sqrt(area)
        return cls(lado)

    @staticmethod
    def es_valido(lado: float) -> bool:
        """Lógica pura: ¿puede existir este cuadrado?"""
        return lado > 0
````

````python

# Uso
cuadrado = Cuadrado.desde_area(100)
print(f"Lado creado: {cuadrado.lado}") # Imprime 10.0
````

````python
cuadrado.es_valido(-5) # True
````


## 4. Métodos Mágicos (Dunder)

**Definición:** Métodos especiales que comienzan y terminan con doble guion bajo (`__`). Permiten que nuestros objetos se comporten como tipos nativos de Python (usar `+`, `print()`, etc.).

**Aplicación en Geometría:** La **Sobrecarga de Operadores**. Si sumas dos vectores, obtienes un nuevo vector. Si imprimes un punto, quieres ver `(x, y)`.




````python
class Vector:
    def __init__(self, x: float, y: float):
        self.x = x
        self.y = y

    def __add__(self, otro: Vector) -> Vector:
        """Sobrecarga del operador + para suma de vectores."""
        return Vector(self.x + otro.x, self.y + otro.y)

    def __str__(self) -> str:
        """Representación amigable para humanos."""
        return f"V({self.x}, {self.y})"

    def __repr__(self) -> str:
        """Representación técnica para desarrolladores."""
        return f"Vector(x={self.x}, y={self.y})"
````

````python

# Ejemplo ilustrativo de geometría analítica
v1 = Vector(1, 0)
v2 = Vector(0, 2)
v3 = v1 + v2  # ¡Esto llama internamente a __add__!
print(v3)     # ¡Esto llama internamente a __str__! (Imprime V(6, 8))
````

````python
v3.__repr__()  # Llama a __repr__, útil para debugging (Imprime Vector(x=6, y=8))
````

### Ejemplo Base: Gestión de Lotes Agropecuarios


````python
class LoteAgricola:
    def __init__(self, id_lote: str, area_ha: float, cultivo: str):
        self.id_lote = id_lote
        self.__area_ha = area_ha  # Privado
        self.cultivo = cultivo
        self.historico_cosechas = []

    @property
    def area(self) -> float:
        return self.__area_ha

    @area.setter
    def area(self, valor: float):
        if valor <= 0:
            raise ValueError("El área debe ser positiva.")
        self.__area_ha = valor

    def registrar_cosecha(self, toneladas: float):
        rendimiento = toneladas / self.__area_ha
        self.historico_cosechas.append(rendimiento)
        print(f"Cosecha registrada. Rendimiento: {rendimiento:.2f} Ton/Ha")

    def __str__(self):
        return f"Lote {self.id_lote} - Cultivo: {self.cultivo} ({self.__area_ha} Ha)"
````

````python

# Uso:
lote1 = LoteAgricola("A-01", 15, "Maíz") # instancia la clase 
#con composición (tiene un área y un tipo de cultivo)
````

````python
lote1.registrar_cosecha(75.0) # registra una cosecha de 75 toneladas
print(lote1)
````

Este es el ejercicio fundamental para entender cómo un **Python Engineer** modela la realidad. En administración y agroindustria, automatizar la nómina es el "Hola Mundo" de los sistemas de gestión.

### Código de la Solución

Copia esto en una celda de tu Jupyter Notebook para analizarlo:

````python
class Empleado:
    # Atributos de clase (Constantes de negocio)
    SALUD_PENSION_PCT: float = 0.08  # 4% salud + 4% pensión
    UVT_MINIMO: float = 2_000_000.0     # Umbral para otras deducciones
    # UVT significa "Unidad de Valor Tributario", es una medida usada en Colombia 
    # para determinar impuestos y deducciones.
    def __init__(self, nombre: str, salario_base: float):
        self.nombre = nombre
        self.salario_base = salario_base

    def calcular_nomina(self) -> float:
        """Calcula el salario neto tras deducciones legales."""
        deduccion_legal = self.salario_base * self.SALUD_PENSION_PCT
        
        # Lógica de negocio: deducción extra si gana más del umbral
        retencion = 0.0
        if self.salario_base > self.UVT_MINIMO:
            retencion = self.salario_base * 0.01  # 1% adicional
            
        salario_neto = self.salario_base - deduccion_legal - retencion
        return salario_neto
````

````python

# --- PRUEBA DEL OBJETO ---
empleado_admin = Empleado("Marco", 2_500_000.0)
pago_final = empleado_admin.calcular_nomina()

print(f"Empleado: {empleado_admin.nombre}")
print(f"Salario Neto a pagar: ${pago_final:,.2f}")
````

## 1. Solución del Problema: Clase Producto

Vamos a modelar un producto donde el `precio_venta` no es un dato estático, sino una **derivación lógica** basada en el costo y el margen de ganancia.



### Opción A: Sin usar `@property` (Forma rudimentaria)
En esta opción, el precio se calcula una sola vez al inicio o requiere llamar a una función cada vez.



````python
class ProductoSimple:
    def __init__(self, nombre: str, costo: float, margen: float):
        self.nombre = nombre
        self.costo = costo
        self.margen = margen
        # El precio se calcula solo al crear el objeto
        self.precio_venta = costo * (1 + margen) 
        # aquí el margen se da en unidades decimales (0.20 para 20%)
````

````python

# Problema:
p1 = ProductoSimple("Fertilizante", 100.0, 0.20)
print(f"Precio inicial: {p1.precio_venta}") # 120.0

p1.costo = 150.0  # El costo sube por la inflación
print(f"Precio tras subida de costo: {p1.precio_venta}") # ¡Sigue siendo 120.0! ERROR.
````

### Opción B: Utilizando `@property` (Forma de Ingeniero)
Aquí el `precio_venta` se comporta como un atributo, pero se calcula en tiempo real cada vez que se consulta.


````python
class ProductoPro:
    def __init__(self, nombre: str, costo: float, margen: float):
        self.nombre = nombre
        self._costo = costo
        self._margen = margen

    @property
    def precio_venta(self) -> float:
        """Calcula el precio dinámicamente: Costo * (1 + Margen)"""
        return self._costo * (1 + self._margen)

    @property
    def costo(self) -> float:
        return self._costo

    @costo.setter
    def costo(self, nuevo_costo: float):
        if nuevo_costo < 0:
            raise ValueError("Un ingeniero no permite costos negativos.")
        self._costo = nuevo_costo
    
````

````python

# Ventaja:
p2 = ProductoPro("Semillas de Maíz", 100.0, 0.20)
print(f"Precio dinámico: {p2.precio_venta}") # 120.0
````

````python

p2.costo = 200.0
print(f"Precio actualizado automáticamente: {p2.precio_venta}") # 180.0
````

| Característica | Sin `@property` (Atributo simple) | Con `@property` (Propiedad) |
| :--- | :--- | :--- |
| **Sincronización** | Los datos pueden quedar desactualizados si uno cambia. | Los datos calculados siempre están al día. |
| **Validación** | Cualquiera puede poner `costo = -500` y romper el sistema. | Puedes usar un `.setter` para proteger el dato. |

| **Refactorización** | Si cambias la lógica, debes buscar en todo tu código dónde se usa. | Cambias la lógica dentro de la clase y el resto del código sigue igual. |

## 3. El sentido de `@property` para un Python Engineer

Para alguien que se forma como ingeniero, `@property` tiene un sentido profundo basado en el **Principio de Acceso Uniforme**.





### A. Mantenibilidad (El código no se rompe)


Imagina que tienes un sistema contable con 10,000 líneas de código que usan `producto.precio_venta`. 

Si el día de mañana la empresa decide que el precio debe incluir un impuesto del 5% adicional, con `@property` solo cambias **una línea** dentro de la clase. Si usaras atributos simples, tendrías que buscar y modificar las 10,000 líneas.


Para actualizar la clase, simplemente debemos modificar la lógica interna de la función decorada con `@property`. Lo interesante aquí es que **el mundo exterior no se entera del cambio**: cualquier código que antes usaba `producto.precio_venta` seguirá funcionando igual, pero ahora recibirá el valor con el impuesto incluido.

Aquí tienes la versión actualizada:

```python


````python
class ProductoPro:
    # Definimos el impuesto como una constante de clase para que sea fácil de mantener
    IMPUESTO_VENTA: float = 0.05 

    def __init__(self, nombre: str, costo: float, margen: float):
        self.nombre = nombre
        self._costo = costo
        self._margen = margen

    @property
    def precio_venta(self) -> float:
        """
        Calcula el precio dinámicamente: (Costo * (1 + Margen)) * (1 + Impuesto)
        """
        precio_antes_impuesto = self._costo * (1 + self._margen)
        precio_final = precio_antes_impuesto * (1 + self.IMPUESTO_VENTA)
        return round(precio_final, 2) # Redondeamos a 2 decimales para manejo de dinero

    @property
    def costo(self) -> float:
        return self._costo

    @costo.setter
    def costo(self, nuevo_costo: float):
        if nuevo_costo < 0:
            raise ValueError("Un ingeniero no permite costos negativos.")
        self._costo = nuevo_costo

    @property
    def margen(self) -> float:
        return self._margen

    @margen.setter
    def margen(self, nuevo_margen: float):
        if not (0 <= nuevo_margen <= 1):
            raise ValueError("El margen debe estar entre 0 y 1 (ej: 0.20 para 20%).")
        self._margen = nuevo_margen
````

````python

# --- Ejemplo de uso ---
p1 = ProductoPro("Fertilizante Urea", 100.0, 0.20)
# Explicación del cálculo: (100 * 1.20) = 120 -> 120 * 1.05 = 126
````

````python
print(f"Producto: {p1.nombre}")
````

````python
print(f"Precio de venta (IVA incluido): ${p1.precio_venta}")
````

### ¿Por qué esto es "Nivel Ingeniero"?

1.  **Uso de Constantes (`IMPUESTO_VENTA`):** No escribimos el `0.05` directamente en la fórmula (eso se llama *hardcoding* y es una mala práctica). Si mañana el gobierno sube el impuesto al **7%**, solo cambias una línea al principio de la clase.
2.  **Encapsulamiento total:** He añadido un `@setter` para el `margen`. Esto asegura que nadie ponga un margen de `-50` o `2000`, protegiendo la lógica de negocio de tu aplicación.
3.  **Transparencia:** Si tenías un reporte de ventas que imprimía miles de precios, al actualizar esta clase **todos los reportes se actualizaron automáticamente**. No tuviste que ir archivo por archivo sumando el 5%.



### Visualizando el flujo de datos con `@property`





Como ves en el diagrama, `@property` actúa como una **aduana**: cuando el programa pide el `precio_venta`, la "aduana" intercepta la petición, realiza el cálculo matemático (incluyendo el nuevo impuesto) y entrega el resultado final.



### B. Validación y "Gobernanza" de Datos
Como ingeniero, no confías en el usuario (ni en otros programadores). Al usar propiedades, puedes interceptar el momento en que alguien intenta cambiar un valor:
* ¿Alguien quiere poner un margen del 5000%? El `@setter` puede lanzar una alerta.
* ¿Alguien quiere cambiar el nombre del producto a un texto vacío? El `@setter` lo impide.



### C. Atributos de "Solo Lectura"
En administración, hay datos que no deben ser modificados manualmente (como el ID de una factura o un saldo calculado). Si creas una `@property` pero **no** le creas un `@setter`, ese atributo se vuelve de "solo lectura". Si alguien intenta hacer `objeto.saldo = 0`, Python lanzará un error, protegiendo la integridad financiera de tu software.



### D. Cómputo bajo demanda (Lazy Evaluation)
Si el cálculo del precio fuera extremadamente complejo (por ejemplo, consultando TRM del dólar en tiempo real), `@property` te permite que ese cálculo solo ocurra cuando realmente alguien pregunta por el precio, ahorrando recursos de procesamiento.



**¿Ves cómo `@property` transforma una simple variable en un "componente inteligente" de tu arquitectura?** 

¿Te gustaría que aplicáramos este concepto a un caso de **Geometría Analítica**, por ejemplo, para que el área de un círculo se actualice sola al cambiar el radio?

````python
class Circulo:
    def __init__(self, radio: float):
        self.__radio = radio  # Atributo privado

    @property
    def radio(self) -> float:
        return self.__radio

    @radio.setter
    def radio(self, valor: float):
        if valor < 0:
            raise ValueError("En geometría euclidiana, el radio no puede ser negativo.")
        self.__radio = valor

    @property
    def area(self) -> float:
        return math.pi * (self.__radio ** 2)
````

````python
c1.radio = -1
print(f"Área del círculo con radio {c1.radio}: {c1.area:.2f}")
````

# 4.  **Clase Cultivo:** Atributos: tipo, ciclo_dias, agua_requerida_litros.


````python
# 4.  **Clase Cultivo:** Atributos: tipo, ciclo_dias, agua_requerida_litros.

class Cultivo:
    def __init__(self, tipo: str, ciclo_dias: int, agua_requerida_litros: float):
        self.tipo = tipo
        self.ciclo_dias = ciclo_dias
        self.agua_requerida_litros = agua_requerida_litros
    @property
    def eficiencia_agua(self) -> float:
        """Calcula la eficiencia del agua como litros por día de ciclo."""
        return self.agua_requerida_litros / self.ciclo_dias
    
    @property
    def rendimiento_estimado(self) -> float:
        """Ejemplo de lógica de negocio: rendimiento estimado basado en eficiencia."""
        if self.eficiencia_agua < 10:
            return 1000  # kg/ha
        elif self.eficiencia_agua < 20:
            return 800
        else:
            return 500
        
        
    def __str__(self):
        return f"Cultivo: {self.tipo}, Ciclo: {self.ciclo_dias} días, Agua requerida: {self.agua_requerida_litros} litros"
````

````python
maiz = Cultivo("Maíz", 120, 1500)
str(maiz)
````

````python
maiz.agua_requerida_litros, maiz.eficiencia_agua, maiz.rendimiento_estimado
````

# 21. **AnalizadorPrecios:** Clase que reciba una lista de precios y calcule media y desviación.


````python
# 21. **AnalizadorPrecios:** Clase que reciba una lista de precios y calcule media y desviación.
from typing import List

class AnalizadorPrecios:
    def __init__(self, precios: List[float]):
        self.precios = precios

    def media(self) -> float:
        return sum(self.precios) / len(self.precios)

    def desviacion_estandar(self) -> float:
        media = self.media()
        varianza = sum((p - media) ** 2 for p in self.precios) / len(self.precios)
        return math.sqrt(varianza)
````

````python

# Ejemplo de uso:
precios = [100.0, 150.0, 200.0, 250.0, 300.0]
analizador = AnalizadorPrecios(precios)  

print(f"Media de precios: {analizador.media():.2f}")
print(f"Desviación estándar: {analizador.desviacion_estandar():.2f}")    
````
