### Celda 1: Introducción y Type Hinting


```python
# Un Python Engineer escribe código claro. El 'Type Hinting' ayuda a documentar y prevenir errores.
from typing import List, Dict, Tuple, Optional

def procesar_usuarios(usuarios: List[str]) -> int:
    """Recibe una lista de nombres y retorna la cantidad procesada."""
    return len(usuarios)

nombres: List[str] = ["Marco", "Elena", "Lucas"]
print(f"Usuarios cargados: {procesar_usuarios(nombres)}")

```


### Celda 2: List Comprehensions (Elegancia y Velocidad)


```python
# En lugar de usar bucles 'for' tradicionales para crear listas, usamos comprensiones.
precios_usd: List[float] = [10.5, 25.0, 100.2, 5.75]

# Supongamos que el dólar hoy está a 4200 COP
tasa_cambio: int = 4200

# Comprensión de lista con condición
precios_cop: List[float] = [p * tasa_cambio for p in precios_usd if p > 10]

print(f"Precios mayores a 10 USD convertidos a COP: {precios_cop}")
```

```python
precios_usd: List[float] = [10.5, 25.0, 100.2, 5.75]
tasa_cambio: int = 3600
precios_cop: List[float] = [p * tasa_cambio for p in precios_usd if p > 10]
print(f"Precios mayores a 10 USD convertidos a COP: {precios_cop}")
```


### Celda 3: Diccionarios y Sets (Búsquedas de alta velocidad)


```python
# Los diccionarios son clave en Python. El método .get() evita errores de ejecución.
configuracion: Dict[str, any] = {
    "db_host": "localhost",
    "puerto": 5432,
    "debug": True
}

# Uso seguro de diccionarios
db_name = configuracion.get("db_host", "default_db") # Si no existe, devuelve el default
db_name
```

```python

# Sets para eliminar duplicados de inmediato
correos: List[str] = ["marco@udea.edu", "test@mail.com", "marco@udea.edu"]
correos_unicos: set = set(correos)

print(f"Diccionario: {configuracion}")
print(f"Correos únicos: {correos_unicos}")
```

### Celda 4: Ejercicio de Práctica


```python
# TODO: Crea una función que reciba una lista de números y devuelva un diccionario 
# donde la clave sea el número y el valor sea su cuadrado, 
# pero solo para los números que sean pares.

def cuadrado_de_pares(numeros: List[int]) -> Dict[int, int]:
    # Tu código aquí
    return {n: n**2 for n in numeros if n % 2 == 1}

# Prueba tu función
print(cuadrado_de_pares([1, 2, 3, 4, 5, 6]))
```
